## Support Vector Machine (SVM) Classification

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

### Step 2: Load Dataset

```python
data=load_breast_cancer()

x=pd.DataFrame(data.data,columns=data.feature_names)

y=pd.Series(data.target)

print("Data Shape:",x.shape)

print(x.head(10))
```

### Step 4: Check Missing Values

```python
print("\nMissing Values")
print(x.isnull().sum())
```

```python
x.describe()
```

### Step 5: Train Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("\nTraining Data Shape:", x_train.shape)
print("Testing Data Shape:", x_test.shape)
```

```python
x.columns
```

### Step 6: Feature Scaling

```python
scaler=StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled =scaler.transform(x_test)
```

### Step 7: Create SVM Model


```python
svm_model=SVC(
    kernel='linear',
    C=1.0,
    random_state=42
)
```

```python
print(x_train_scaled)
```

### Step 8: Train Model

```python
svm_model.fit(x_train_scaled, y_train)

print("\nModel Training Completed")
```

### Step 9: Predictions

```python
y_pred = svm_model.predict(x_test_scaled)

print("\nPredictions")
print(y_pred[:10])
```

### Step 10: Accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy Score")
print(round(accuracy,4))
```

### Step 11: Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix")
print(cm)
```

```python
#Heatmap
import seaborn as sns
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='inferno')
plt.xlabel("Predicted")
plt.ylabel("Confusion Matrix")
plt.show()
```

### Step 12: Classification Report

```python
report=classification_report(y_test,y_pred)

print("\nClassification report\n",report)
```

### Step 13: RBF Kernel SVM

```python
rbf_model=SVC(
    kernel='rbf',
    C=0.1,
    gamma='scale',
    random_state=42
)
rbf_model.fit(x_train_scaled,y_train)
rbf_pred=rbf_model.predict(x_test_scaled)
rbf_accuracy=accuracy_score(y_test,rbf_pred)

print("\nRBF Kernel Accuracy")
print(round(rbf_accuracy,4))
```

### Step 14: Compare Models

```python
print("\nModel Comparison")
print("Linear Kernel Accuracy :", accuracy)
print("RBF Kernel Accuracy    :", rbf_accuracy)
```

### Step 15: Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf']
}

grid_search=GridSearchCV(
    SVC(),
    param_grid,
    cv=5,
    verbose=1
)
grid_search.fit(x_train_scaled, y_train)

print("\nBest Parameters")
print(grid_search.best_params_)
```

### Step 16: Best Model Evaluation

```python
best_model = grid_search.best_estimator_

best_pred = best_model.predict(x_test_scaled)

best_accuracy = accuracy_score(y_test, best_pred)

print("\nBest Model Accuracy")
print(best_accuracy)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    best_model,
    x_train_scaled,
    y_train,
    cv=5
)

print("\nCross Validation Scores")
print(scores)

print("\nAverage CV Score")
print(scores.mean())
```
